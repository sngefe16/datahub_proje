# Exploratory Data Analysis - IEEE Fraud Detection

This notebook performs initial exploration of the IEEE Fraud Detection dataset.


```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Import project modules
import sys
sys.path.append('../')
from src.data_loader import load_data, get_feature_types, get_missing_stats
from src.preprocessing import reduce_memory_usage
```

## 1. Load Data


```python
# Load data (use sample_size for faster development)
train_df, test_df = load_data(data_dir='../data', sample_size=None, use_identity=True)

print(f"Training shape: {train_df.shape}")
print(f"Test shape: {test_df.shape}")
```

## 2. Basic Statistics & Target Analysis


```python
# Target variable analysis
print("Target Distribution:")
print(train_df['isFraud'].value_counts())
print(f"\nFraud Rate: {train_df['isFraud'].mean():.4f}")

# Visualize target distribution
plt.figure(figsize=(8, 5))
train_df['isFraud'].value_counts().plot(kind='bar')
plt.title('Target Variable Distribution')
plt.xlabel('isFraud')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

## 3. Missing Values Analysis


```python
# Missing value statistics
missing_stats = get_missing_stats(train_df)

print("Top 20 Features with Missing Values:")
print(missing_stats.head(20))

# Visualize missing values
plt.figure(figsize=(12, 8))
missing_stats.head(30)['missing_pct'].plot(kind='barh')
plt.title('Top 30 Features by Missing Percentage')
plt.xlabel('Missing Percentage (%)')
plt.ylabel('Feature')
plt.show()
```

## 4. Transaction Amount Analysis


```python
# Transaction amount statistics
print("Transaction Amount Statistics:")
print(train_df['TransactionAmt'].describe())

# Distribution by fraud
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Histogram
train_df[train_df['isFraud'] == 0]['TransactionAmt'].hist(bins=50, ax=axes[0], alpha=0.7, label='Not Fraud')
train_df[train_df['isFraud'] == 1]['TransactionAmt'].hist(bins=50, ax=axes[0], alpha=0.7, label='Fraud')
axes[0].set_title('Transaction Amount Distribution')
axes[0].set_xlabel('Transaction Amount')
axes[0].set_ylabel('Frequency')
axes[0].legend()

# Log scale
train_df[train_df['isFraud'] == 0]['TransactionAmt'].apply(np.log1p).hist(bins=50, ax=axes[1], alpha=0.7, label='Not Fraud')
train_df[train_df['isFraud'] == 1]['TransactionAmt'].apply(np.log1p).hist(bins=50, ax=axes[1], alpha=0.7, label='Fraud')
axes[1].set_title('Transaction Amount Distribution (Log Scale)')
axes[1].set_xlabel('Log(Transaction Amount)')
axes[1].set_ylabel('Frequency')
axes[1].legend()

plt.tight_layout()
plt.show()
```
